Create an interative Python script that will use the attached CSV file for it's data load. Loading that data into a pandas dataframe, create 2 linear regression models, and graph each one using MatPlotLib. Show the assessment statistics (r^2, p-value, slope, and intercept) for each model. Use the two models to make the indicated predictions. Based on the assessments and predictions, add an explanation of if each independent variable is statistically significant to the target variable, and how much variation in the target variable the model you have built explains. Try to put this as "real world" as you can.



1. Your first model should use "Horsepower" to predict "MPG". Show a visualization of the relationship between these two fields. Predict the MPG of a car with 125 Horsepower.

2. Your second model should use "Year" to predict "MPG". Show a visualization of the relationship between these two fields. Predict what MPG a car built in the year 1984 would be estimated as using this data.


```python
import pandas as pd
from scipy import stats
import seaborn as sns
import scipy.stats as stats
mpg = pd.read_csv('AutoMPGdata.csv')

#1
linear_regression1 = stats.linregress(x=mpg.Horsepower, y=mpg.MPG)
sns.set_style('whitegrid')
axes = sns.regplot(x=mpg.Horsepower, y=mpg.MPG)

print (f'r^2 value: {linear_regression1.rvalue**2:.2f}') 
print (f'p-value: {linear_regression1.pvalue:.80f}*') 
print(f'slope: {linear_regression1.slope:.2f}')
print(f'intercept: {linear_regression1.intercept:.2f}')

print (f"Predicted MPG of a car with 125 Horsepower: {linear_regression1.slope * 125 + linear_regression1.intercept:.1f}")
```

Explanation 1:

There is a moderately strong relationship between Horsepower and MPG (r^2 = 0.61), which means Horsepower is statistically significant in predicting MPG (p < 0.05). For the real world, this shows that Horsepower is a major factor in determining the fuel efficiency of a car. This makes sense: a higher horsepower engine requires more energy to operate, and therefore results in lower fuel efficiency.

```python
#2
linear_regression2 = stats.linregress(x=mpg.Year, y=mpg.MPG)
sns.set_style('whitegrid')
axes2 = sns.regplot(x=mpg.Year, y=mpg.MPG)

print (f'r^2 value: {linear_regression2.rvalue**2:.2f}') 
print (f'p-value: {linear_regression2.pvalue:.36f}*') 
print(f'slope: {linear_regression2.slope:.2f}')
print(f'intercept: {linear_regression2.intercept:.2f}')

print (f"Predicted MPG of a car built in 1984: {(linear_regression2.slope * 84 + linear_regression2.intercept):.1f}")
```

Explanation 2:

There is a moderately weak relationship between the year a car was made and MPG (r^2 = 0.34), but Year is still statistically significant in predciting MPG (p < 0.05). For the real world, the year is likely to effect MPG, since newer cars would contain more advanced technology that advance fuel efficiency. However, there are more factors to consider that contribute more to fuel efficiency than the year it was made.
